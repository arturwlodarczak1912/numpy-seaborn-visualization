# Wektoryzacja w numpy

```python
def sondaz(N):
  Trump = 0
  Kamala = 0
  for i in range(int(N)):
    vote = random.choice(['Trump','Kamala'])
    if vote == 'Trump':
      Trump += 1
    else:
      Kamala += 1
  return abs(Trump - Kamala)/N
```

```python
sondaz(10)
```

```python
N = 800
```

```python
vec_calculations = np.vectorize(sondaz)
```

```python
np.ones((100, 1))*800
```

```python
result = vec_calculations(np.ones((100, 1))*800)  #  dzięki wektoryzacji nie musze pisać for, i kod będzie działał szybko
```

```python
result.shape
```

```python
def run_numpy(N):
  return vec_calculations(np.ones((N, 1))*800)
```

```python
def list_comp(N):
  return [sondaz(800) for i in range(N)]
```

```python
list_comp(2)
```

```python
%timeit run_numpy(1000)
```

```python
%timeit list_comp(1000)
```
